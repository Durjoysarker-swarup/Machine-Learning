```python
# Importing libraries
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Building the corpus
corpus = [
    "The news is fake",
    "The news is real",
    "Fake news is bad"
]
```

```python
# Initialize vectorizer
vectorizer = TfidfVectorizer()
```

```python
# Fit and transform
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python
print(tfidf_matrix)
```

```python
tfidf_arrary = tfidf_matrix.toarray()
```

```python
print(tfidf_arrary)
```

```python
feature_name = vectorizer.get_feature_names_out()
```

```python
print(feature_name)
```

```python
# Creating a  readable table
df_tfidf = pd.DataFrame(tfidf_arrary, columns=feature_name)
```

```python
print(df_tfidf)
```
