# Fake News Detection

### Load Dataset

```python
import pandas as pd
df_true = pd.read_csv('True.csv')
df_fake = pd.read_csv('Fake.csv')
```

```python
#can check the dataset
#df_true.head()
#df_true.shape
#df_fake.head()
#df_fake.shape
```

### Add labels: 1 for real, 0 for fake

```python
df_true["label"] = 1
df_fake["label"] = 0
```

### Combine and Shuffle

#### Combine

```python
df = pd.concat([df_true,df_fake])
```

```python
df.to_csv("combined_data.csv", index=False)
```
